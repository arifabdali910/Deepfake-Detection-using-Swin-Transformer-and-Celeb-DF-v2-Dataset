```python
###############################################################################
#  DEEPFAKE DETECTION — Swin Transformer Tiny Fine-Tuning
#  ------------------------------------------------------
#  • Dataset: ~191K face crops (real vs fake) from Celeb-DF-v2 preprocessing
#  • Model:   swin_tiny_patch4_window7_224 (~28M params, ImageNet pretrained)
#  • Split:   70% train / 15% val / 15% test (stratified, no leakage)
#  • Balance: 2× oversample REAL in TRAINING SET ONLY
#  • Regularization: augmentation, label smoothing, weight decay, drop-path,
#                    class weights, early stopping, gradient clipping, AMP
#  • Outputs everything to /kaggle/working/
#
#  Copy this whole file into a single Kaggle notebook cell, or split it at
#  the `# ==== CELL N ====` markers into separate cells.
###############################################################################


# ============================================================================
# CELL 1 — Install & imports
# ============================================================================
import subprocess, sys
subprocess.check_call(
    [sys.executable, "-m", "pip", "install", "-q", "timm", "--no-warn-script-location"]
)

import os
import gc
import json
import random
import time
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image, ImageFile

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torch.cuda.amp import autocast, GradScaler

import torchvision.transforms as T
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report, roc_curve,
    precision_recall_curve, average_precision_score, roc_auc_score,
)

import timm
from tqdm.auto import tqdm

warnings.filterwarnings("ignore")
ImageFile.LOAD_TRUNCATED_IMAGES = True

print("✅  Libraries loaded")
print(f"   torch  : {torch.__version__}")
print(f"   timm   : {timm.__version__}")


# ============================================================================
# CELL 2 — Configuration
# ============================================================================
class CFG:
    # --- Paths ---
    REAL_DIR   = "//kaggle/input/datasets/irfan3464/celeb-df-irfan/preprocessed_faces/real"
    FAKE_DIR   = "/kaggle/input/datasets/irfan3464/celeb-df-irfan/preprocessed_faces/fake"
    OUTPUT_DIR = "/kaggle/working"
    PLOT_DIR   = "/kaggle/working/plots"

    # --- Model ---
    MODEL_NAME  = "swin_tiny_patch4_window7_224"   # ~28M params
    NUM_CLASSES = 2
    IMAGE_SIZE  = 224

    # --- Training ---
    BATCH_SIZE       = 64
    NUM_EPOCHS       = 150
    LR_HEAD          = 1e-4        # classifier head learning rate
    LR_BACKBONE      = 1e-5        # backbone learning rate (10× smaller)
    WEIGHT_DECAY     = 0.05
    LABEL_SMOOTHING  = 0.1
    GRAD_CLIP        = 1.0
    WARMUP_EPOCHS    = 1
    DROP_RATE        = 0.2         # classifier dropout
    DROP_PATH_RATE   = 0.1         # stochastic depth

    # --- Data ---
    TRAIN_RATIO             = 0.70
    VAL_RATIO               = 0.15
    TEST_RATIO              = 0.15
    REAL_OVERSAMPLE_FACTOR  = 2    # duplicate real 2× (with different augmentations)
    NUM_WORKERS             = 2

    # --- Early stopping ---
    PATIENCE = 5

    # --- Reproducibility ---
    SEED = 42

    # --- ImageNet normalization (Swin pretrained on ImageNet) ---
    MEAN = [0.485, 0.456, 0.406]
    STD  = [0.229, 0.224, 0.225]

os.makedirs(CFG.OUTPUT_DIR, exist_ok=True)
os.makedirs(CFG.PLOT_DIR,   exist_ok=True)


# ============================================================================
# CELL 3 — Reproducibility & device
# ============================================================================
def seed_everything(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = False
    torch.backends.cudnn.benchmark     = True

seed_everything(CFG.SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"🖥  Device: {device}")
if device.type == "cuda":
    print(f"   GPU  : {torch.cuda.get_device_name(0)}")
    print(f"   VRAM : {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")


# ============================================================================
# CELL 4 — Scan real & fake folders → build DataFrame
# ============================================================================
VALID_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

def list_images(folder):
    if not os.path.isdir(folder):
        raise FileNotFoundError(f"Folder not found: {folder}")
    files = [
        os.path.join(folder, f)
        for f in os.listdir(folder)
        if Path(f).suffix.lower() in VALID_EXT
    ]
    return sorted(files)

print("📂  Scanning image folders …")
real_files = list_images(CFG.REAL_DIR)
fake_files = list_images(CFG.FAKE_DIR)

# label 0 = REAL, label 1 = FAKE
df = pd.DataFrame({
    "path":  real_files + fake_files,
    "label": [0] * len(real_files) + [1] * len(fake_files),
})

print(f"   Real images : {len(real_files):,}")
print(f"   Fake images : {len(fake_files):,}")
print(f"   Total       : {len(df):,}")
print(f"   Ratio real:fake = 1 : {len(fake_files)/max(len(real_files),1):.2f}")

if len(df) == 0:
    raise RuntimeError("No images found. Double-check REAL_DIR and FAKE_DIR.")


# ============================================================================
# CELL 5 — Stratified 70 / 15 / 15 split (before any oversampling)
# ============================================================================
train_df, temp_df = train_test_split(
    df, test_size=(CFG.VAL_RATIO + CFG.TEST_RATIO),
    stratify=df["label"], random_state=CFG.SEED,
)
val_df, test_df = train_test_split(
    temp_df, test_size=CFG.TEST_RATIO / (CFG.VAL_RATIO + CFG.TEST_RATIO),
    stratify=temp_df["label"], random_state=CFG.SEED,
)
train_df = train_df.reset_index(drop=True)
val_df   = val_df.reset_index(drop=True)
test_df  = test_df.reset_index(drop=True)

def dist(d): return d["label"].value_counts().sort_index().to_dict()
print(f"   Train : {len(train_df):>7,}   real/fake = {dist(train_df)}")
print(f"   Val   : {len(val_df):>7,}   real/fake = {dist(val_df)}")
print(f"   Test  : {len(test_df):>7,}   real/fake = {dist(test_df)}")


# ============================================================================
# CELL 6 — Oversample REAL by 2× in TRAINING SET ONLY
# ============================================================================
real_train = train_df[train_df["label"] == 0]
fake_train = train_df[train_df["label"] == 1]

# Duplicate real rows REAL_OVERSAMPLE_FACTOR times.
# Since our train_transform applies stochastic augmentation at __getitem__ time,
# each duplicate will look different every epoch → effectively "augmented 2×".
real_oversampled = pd.concat([real_train] * CFG.REAL_OVERSAMPLE_FACTOR, ignore_index=True)

train_df_bal = (
    pd.concat([real_oversampled, fake_train], ignore_index=True)
    .sample(frac=1.0, random_state=CFG.SEED)
    .reset_index(drop=True)
)

print(f"   Training set after {CFG.REAL_OVERSAMPLE_FACTOR}× real oversampling: {len(train_df_bal):,}")
print(f"     real / fake = {dist(train_df_bal)}")


# ============================================================================
# CELL 7 — Transforms (normalization + augmentation)
# ============================================================================
train_transform = T.Compose([
    T.Resize((CFG.IMAGE_SIZE + 16, CFG.IMAGE_SIZE + 16)),
    T.RandomCrop(CFG.IMAGE_SIZE),
    T.RandomHorizontalFlip(p=0.5),
    T.RandomApply([T.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2, hue=0.05)], p=0.5),
    T.RandomApply([T.GaussianBlur(kernel_size=3, sigma=(0.1, 1.5))], p=0.2),
    T.RandomRotation(degrees=10),
    T.ToTensor(),
    T.Normalize(mean=CFG.MEAN, std=CFG.STD),
    T.RandomErasing(p=0.25, scale=(0.02, 0.10)),
])

eval_transform = T.Compose([
    T.Resize(256),
    T.CenterCrop(CFG.IMAGE_SIZE),
    T.ToTensor(),
    T.Normalize(mean=CFG.MEAN, std=CFG.STD),
])


# ============================================================================
# CELL 8 — Dataset + DataLoaders
# ============================================================================
class DeepfakeDataset(Dataset):
    def __init__(self, dataframe, transform=None):
        self.paths     = dataframe["path"].values
        self.labels    = dataframe["label"].values.astype(np.int64)
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, idx):
        path  = self.paths[idx]
        label = int(self.labels[idx])
        try:
            img = Image.open(path).convert("RGB")
        except Exception:
            # Robust fallback for the rare corrupt file — avoids crashing training
            img = Image.new("RGB", (CFG.IMAGE_SIZE, CFG.IMAGE_SIZE), (0, 0, 0))
        if self.transform is not None:
            img = self.transform(img)
        return img, label


train_ds = DeepfakeDataset(train_df_bal, transform=train_transform)
val_ds   = DeepfakeDataset(val_df,       transform=eval_transform)
test_ds  = DeepfakeDataset(test_df,      transform=eval_transform)

train_loader = DataLoader(
    train_ds, batch_size=CFG.BATCH_SIZE, shuffle=True,
    num_workers=CFG.NUM_WORKERS, pin_memory=True, drop_last=True,
    persistent_workers=(CFG.NUM_WORKERS > 0),
)
val_loader = DataLoader(
    val_ds, batch_size=CFG.BATCH_SIZE, shuffle=False,
    num_workers=CFG.NUM_WORKERS, pin_memory=True,
    persistent_workers=(CFG.NUM_WORKERS > 0),
)
test_loader = DataLoader(
    test_ds, batch_size=CFG.BATCH_SIZE, shuffle=False,
    num_workers=CFG.NUM_WORKERS, pin_memory=True,
    persistent_workers=(CFG.NUM_WORKERS > 0),
)
print(f"   Train batches: {len(train_loader)}")
print(f"   Val   batches: {len(val_loader)}")
print(f"   Test  batches: {len(test_loader)}")


# ============================================================================
# CELL 9 — Visualise class distribution & split sizes
# ============================================================================
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Original
orig = df["label"].value_counts().sort_index()
bars = axes[0].bar(["Real", "Fake"], orig.values, color=["#27ae60", "#c0392b"])
axes[0].set_title("Original Full Dataset", fontsize=13, fontweight="bold")
axes[0].set_ylabel("Count")
for b, v in zip(bars, orig.values):
    axes[0].text(b.get_x() + b.get_width() / 2, v, f"{v:,}",
                 ha="center", va="bottom", fontweight="bold")

# Balanced training set
tbal = train_df_bal["label"].value_counts().sort_index()
bars = axes[1].bar(["Real (2×)", "Fake"], tbal.values, color=["#27ae60", "#c0392b"])
axes[1].set_title("Training Set (after 2× real oversampling)", fontsize=13, fontweight="bold")
axes[1].set_ylabel("Count")
for b, v in zip(bars, tbal.values):
    axes[1].text(b.get_x() + b.get_width() / 2, v, f"{v:,}",
                 ha="center", va="bottom", fontweight="bold")

# Split sizes
split_sizes = [len(train_df_bal), len(val_df), len(test_df)]
bars = axes[2].bar(["Train", "Val", "Test"], split_sizes,
                   color=["#3498db", "#f39c12", "#9b59b6"])
axes[2].set_title("Split Sizes", fontsize=13, fontweight="bold")
axes[2].set_ylabel("Count")
for b, v in zip(bars, split_sizes):
    axes[2].text(b.get_x() + b.get_width() / 2, v, f"{v:,}",
                 ha="center", va="bottom", fontweight="bold")

plt.tight_layout()
plt.savefig(os.path.join(CFG.PLOT_DIR, "01_class_distribution.png"), dpi=100, bbox_inches="tight")
plt.show()


# ============================================================================
# CELL 10 — Visualise sample augmented images
# ============================================================================
def denormalize(tensor):
    mean = torch.tensor(CFG.MEAN).view(3, 1, 1)
    std  = torch.tensor(CFG.STD).view(3, 1, 1)
    return torch.clamp(tensor * std + mean, 0, 1)

fig, axes = plt.subplots(2, 6, figsize=(18, 6))
real_shown = fake_shown = 0
for idx in range(len(train_ds)):
    if real_shown >= 6 and fake_shown >= 6:
        break
    img, lbl = train_ds[idx]
    if lbl == 0 and real_shown < 6:
        axes[0, real_shown].imshow(denormalize(img).permute(1, 2, 0).numpy())
        axes[0, real_shown].set_title("Real", color="green", fontweight="bold")
        axes[0, real_shown].axis("off")
        real_shown += 1
    elif lbl == 1 and fake_shown < 6:
        axes[1, fake_shown].imshow(denormalize(img).permute(1, 2, 0).numpy())
        axes[1, fake_shown].set_title("Fake", color="red", fontweight="bold")
        axes[1, fake_shown].axis("off")
        fake_shown += 1

plt.suptitle("Sample training images (after normalization + augmentation)",
             fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig(os.path.join(CFG.PLOT_DIR, "02_sample_images.png"), dpi=100, bbox_inches="tight")
plt.show()


# ============================================================================
# CELL 11 — Build Swin-T model
# ============================================================================
model = timm.create_model(
    CFG.MODEL_NAME,
    pretrained=True,
    num_classes=CFG.NUM_CLASSES,
    drop_rate=CFG.DROP_RATE,
    drop_path_rate=CFG.DROP_PATH_RATE,
).to(device)

total_params     = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"   Model       : {CFG.MODEL_NAME}")
print(f"   Total params    : {total_params/1e6:.2f} M")
print(f"   Trainable params: {trainable_params/1e6:.2f} M")


# ============================================================================
# CELL 12 — Loss, optimizer, scheduler, AMP
# ============================================================================
# Class weights based on the (balanced) training set — small correction only
counts = train_df_bal["label"].value_counts().sort_index().values.astype(np.float32)
class_weights = torch.tensor(len(train_df_bal) / (len(counts) * counts), dtype=torch.float32).to(device)
print(f"   Class weights (CE): real={class_weights[0]:.3f}, fake={class_weights[1]:.3f}")

criterion = nn.CrossEntropyLoss(weight=class_weights, label_smoothing=CFG.LABEL_SMOOTHING)

# Discriminative LR — smaller for the pretrained backbone, larger for the new head
head_params, backbone_params = [], []
for name, p in model.named_parameters():
    if not p.requires_grad:
        continue
    (head_params if "head" in name else backbone_params).append(p)
print(f"   Backbone params: {sum(p.numel() for p in backbone_params)/1e6:.2f} M")
print(f"   Head     params: {sum(p.numel() for p in head_params)/1e6:.2f} M")

optimizer = optim.AdamW(
    [
        {"params": backbone_params, "lr": CFG.LR_BACKBONE},
        {"params": head_params,     "lr": CFG.LR_HEAD},
    ],
    weight_decay=CFG.WEIGHT_DECAY,
)

steps_per_epoch = len(train_loader)
total_steps  = steps_per_epoch * CFG.NUM_EPOCHS
warmup_steps = steps_per_epoch * CFG.WARMUP_EPOCHS

def lr_lambda(step):
    if step < warmup_steps:
        return float(step) / float(max(1, warmup_steps))
    progress = (step - warmup_steps) / float(max(1, total_steps - warmup_steps))
    return 0.5 * (1.0 + np.cos(np.pi * progress))

scheduler = optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)
scaler    = GradScaler() if device.type == "cuda" else None


# ============================================================================
# CELL 13 — Train / evaluate helpers
# ============================================================================
def train_one_epoch(model, loader, criterion, optimizer, scheduler, scaler, device, epoch):
    model.train()
    running_loss = 0.0
    all_preds, all_labels = [], []

    pbar = tqdm(loader, desc=f"Epoch {epoch+1:02d} [train]", leave=False)
    for imgs, labels in pbar:
        imgs   = imgs.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)

        if scaler is not None:
            with autocast():
                logits = model(imgs)
                loss   = criterion(logits, labels)
            scaler.scale(loss).backward()
            scaler.unscale_(optimizer)
            torch.nn.utils.clip_grad_norm_(model.parameters(), CFG.GRAD_CLIP)
            scaler.step(optimizer)
            scaler.update()
        else:
            logits = model(imgs)
            loss   = criterion(logits, labels)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), CFG.GRAD_CLIP)
            optimizer.step()

        scheduler.step()

        running_loss += loss.item() * imgs.size(0)
        preds = logits.argmax(dim=1)
        all_preds.extend(preds.detach().cpu().numpy().tolist())
        all_labels.extend(labels.detach().cpu().numpy().tolist())
        pbar.set_postfix(loss=f"{loss.item():.4f}",
                         lr=f"{optimizer.param_groups[-1]['lr']:.2e}")

    epoch_loss = running_loss / len(loader.dataset)
    acc = accuracy_score(all_labels, all_preds)
    f1  = f1_score(all_labels, all_preds, average="binary")
    return epoch_loss, acc, f1


@torch.no_grad()
def evaluate(model, loader, criterion, device, desc="val"):
    model.eval()
    running_loss = 0.0
    all_preds, all_labels, all_probs = [], [], []

    for imgs, labels in tqdm(loader, desc=f"[{desc}]", leave=False):
        imgs   = imgs.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)

        if device.type == "cuda":
            with autocast():
                logits = model(imgs)
                loss   = criterion(logits, labels)
        else:
            logits = model(imgs)
            loss   = criterion(logits, labels)

        probs = torch.softmax(logits.float(), dim=1)[:, 1]
        preds = logits.argmax(dim=1)

        running_loss += loss.item() * imgs.size(0)
        all_preds.extend(preds.cpu().numpy().tolist())
        all_labels.extend(labels.cpu().numpy().tolist())
        all_probs.extend(probs.cpu().numpy().tolist())

    epoch_loss = running_loss / len(loader.dataset)
    acc = accuracy_score(all_labels, all_preds)
    f1  = f1_score(all_labels, all_preds, average="binary")
    return epoch_loss, acc, f1, np.array(all_labels), np.array(all_preds), np.array(all_probs)


# ============================================================================
# CELL 14 — Training loop  (best-model saving + early stopping)
# ============================================================================
history = {"train_loss": [], "train_acc": [], "train_f1": [],
           "val_loss":   [], "val_acc":   [], "val_f1":   [],
           "lr":         []}

best_val_f1        = -1.0
best_epoch         = -1
epochs_no_improve  = 0
best_ckpt_path     = os.path.join(CFG.OUTPUT_DIR, "best_model.pth")
final_ckpt_path    = os.path.join(CFG.OUTPUT_DIR, "final_model.pth")

print("=" * 70)
print(f"  🚀  Starting training — {CFG.NUM_EPOCHS} epochs max, patience={CFG.PATIENCE}")
print("=" * 70)
t_start = time.time()

for epoch in range(CFG.NUM_EPOCHS):
    ep_start = time.time()

    tr_loss, tr_acc, tr_f1 = train_one_epoch(
        model, train_loader, criterion, optimizer, scheduler, scaler, device, epoch
    )
    va_loss, va_acc, va_f1, _, _, _ = evaluate(
        model, val_loader, criterion, device, desc=f"ep {epoch+1:02d} val"
    )

    history["train_loss"].append(tr_loss); history["train_acc"].append(tr_acc); history["train_f1"].append(tr_f1)
    history["val_loss"].append(va_loss);   history["val_acc"].append(va_acc);   history["val_f1"].append(va_f1)
    history["lr"].append(optimizer.param_groups[-1]["lr"])

    ep_time = time.time() - ep_start
    print(f"\nEpoch {epoch+1:02d}/{CFG.NUM_EPOCHS}  ({ep_time/60:.1f} min)")
    print(f"  Train : loss={tr_loss:.4f}  acc={tr_acc:.4f}  f1={tr_f1:.4f}")
    print(f"  Val   : loss={va_loss:.4f}  acc={va_acc:.4f}  f1={va_f1:.4f}")

    # --- Save best model on val F1 ---
    if va_f1 > best_val_f1:
        best_val_f1       = va_f1
        best_epoch        = epoch + 1
        epochs_no_improve = 0
        torch.save({
            "epoch":            epoch + 1,
            "model_state_dict": model.state_dict(),
            "val_f1":           va_f1,
            "val_acc":          va_acc,
            "model_name":       CFG.MODEL_NAME,
            "num_classes":      CFG.NUM_CLASSES,
        }, best_ckpt_path)
        print(f"  ✅  New best model saved  (val F1 = {va_f1:.4f})")
    else:
        epochs_no_improve += 1
        print(f"  ⚠  no improvement ({epochs_no_improve}/{CFG.PATIENCE})")

    # persist history each epoch
    with open(os.path.join(CFG.OUTPUT_DIR, "training_history.json"), "w") as f:
        json.dump(history, f, indent=2)

    if epochs_no_improve >= CFG.PATIENCE:
        print(f"\n⏹  Early stopping triggered at epoch {epoch+1}")
        break

# Save final model as well (post-last-epoch weights, not necessarily best)
torch.save({
    "epoch":            len(history["train_loss"]),
    "model_state_dict": model.state_dict(),
    "model_name":       CFG.MODEL_NAME,
    "num_classes":      CFG.NUM_CLASSES,
}, final_ckpt_path)

total_time = time.time() - t_start
print("\n" + "=" * 70)
print(f"  ✅  Training done in {total_time/60:.1f} min")
print(f"     Best epoch : {best_epoch}")
print(f"     Best val F1: {best_val_f1:.4f}")
print(f"     Saved      : {best_ckpt_path}")
print("=" * 70)


# ============================================================================
# CELL 15 — Training curves
# ============================================================================
epochs_ran = list(range(1, len(history["train_loss"]) + 1))
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

axes[0, 0].plot(epochs_ran, history["train_loss"], "b-o", label="Train")
axes[0, 0].plot(epochs_ran, history["val_loss"],   "r-o", label="Val")
axes[0, 0].axvline(best_epoch, color="g", linestyle="--", alpha=0.6, label=f"Best (ep {best_epoch})")
axes[0, 0].set_title("Loss", fontweight="bold")
axes[0, 0].set_xlabel("Epoch"); axes[0, 0].set_ylabel("Loss")
axes[0, 0].legend(); axes[0, 0].grid(alpha=0.3)

axes[0, 1].plot(epochs_ran, history["train_acc"], "b-o", label="Train")
axes[0, 1].plot(epochs_ran, history["val_acc"],   "r-o", label="Val")
axes[0, 1].axvline(best_epoch, color="g", linestyle="--", alpha=0.6)
axes[0, 1].set_title("Accuracy", fontweight="bold")
axes[0, 1].set_xlabel("Epoch"); axes[0, 1].set_ylabel("Accuracy")
axes[0, 1].legend(); axes[0, 1].grid(alpha=0.3)

axes[1, 0].plot(epochs_ran, history["train_f1"], "b-o", label="Train")
axes[1, 0].plot(epochs_ran, history["val_f1"],   "r-o", label="Val")
axes[1, 0].axvline(best_epoch, color="g", linestyle="--", alpha=0.6)
axes[1, 0].set_title("F1 Score", fontweight="bold")
axes[1, 0].set_xlabel("Epoch"); axes[1, 0].set_ylabel("F1")
axes[1, 0].legend(); axes[1, 0].grid(alpha=0.3)

axes[1, 1].plot(epochs_ran, history["lr"], "purple", marker="o")
axes[1, 1].set_title("Learning Rate (head)", fontweight="bold")
axes[1, 1].set_xlabel("Epoch"); axes[1, 1].set_ylabel("LR")
axes[1, 1].set_yscale("log"); axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(CFG.PLOT_DIR, "03_training_curves.png"), dpi=100, bbox_inches="tight")
plt.show()


# ============================================================================
# CELL 16 — Load best model & evaluate on TEST set
# ============================================================================
print("📥  Loading best checkpoint …")
ckpt = torch.load(best_ckpt_path, map_location=device, weights_only=False)
model.load_state_dict(ckpt["model_state_dict"])
print(f"   Loaded epoch {ckpt['epoch']}  (val F1 = {ckpt['val_f1']:.4f})")

test_loss, test_acc, test_f1, y_true, y_pred, y_prob = evaluate(
    model, test_loader, criterion, device, desc="test"
)

test_precision = precision_score(y_true, y_pred, zero_division=0)
test_recall    = recall_score(y_true, y_pred, zero_division=0)
test_auc       = roc_auc_score(y_true, y_prob)
test_ap        = average_precision_score(y_true, y_prob)

print("\n" + "=" * 70)
print("  📊  TEST SET METRICS")
print("=" * 70)
print(f"  Loss      : {test_loss:.4f}")
print(f"  Accuracy  : {test_acc:.4f}")
print(f"  Precision : {test_precision:.4f}")
print(f"  Recall    : {test_recall:.4f}")
print(f"  F1        : {test_f1:.4f}")
print(f"  ROC-AUC   : {test_auc:.4f}")
print(f"  PR-AUC    : {test_ap:.4f}")
print("\n" + classification_report(y_true, y_pred, target_names=["Real", "Fake"]))

test_metrics = {
    "loss": test_loss, "accuracy": test_acc,
    "precision": test_precision, "recall": test_recall,
    "f1": test_f1, "roc_auc": test_auc, "pr_auc": test_ap,
    "classification_report": classification_report(
        y_true, y_pred, target_names=["Real", "Fake"], output_dict=True, zero_division=0
    ),
    "best_epoch": int(best_epoch),
    "best_val_f1": float(best_val_f1),
}
with open(os.path.join(CFG.OUTPUT_DIR, "test_metrics.json"), "w") as f:
    json.dump(test_metrics, f, indent=2)


# ============================================================================
# CELL 17 — Confusion matrix + ROC + PR curves
# ============================================================================
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

cm = confusion_matrix(y_true, y_pred)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Real", "Fake"], yticklabels=["Real", "Fake"], ax=axes[0], cbar=False)
axes[0].set_title(f"Confusion Matrix  (Acc {test_acc:.4f})", fontweight="bold")
axes[0].set_xlabel("Predicted"); axes[0].set_ylabel("Actual")

fpr, tpr, _ = roc_curve(y_true, y_prob)
axes[1].plot(fpr, tpr, color="darkorange", lw=2, label=f"AUC = {test_auc:.4f}")
axes[1].plot([0, 1], [0, 1], color="navy", lw=1, linestyle="--")
axes[1].set_xlabel("False Positive Rate"); axes[1].set_ylabel("True Positive Rate")
axes[1].set_title("ROC Curve", fontweight="bold")
axes[1].legend(loc="lower right"); axes[1].grid(alpha=0.3)

prec, rec, _ = precision_recall_curve(y_true, y_prob)
axes[2].plot(rec, prec, color="darkgreen", lw=2, label=f"AP = {test_ap:.4f}")
axes[2].set_xlabel("Recall"); axes[2].set_ylabel("Precision")
axes[2].set_title("Precision-Recall Curve", fontweight="bold")
axes[2].legend(loc="lower left"); axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.savefig(os.path.join(CFG.PLOT_DIR, "04_test_metrics.png"), dpi=100, bbox_inches="tight")
plt.show()

# Normalized CM
fig, ax = plt.subplots(figsize=(6, 5))
cm_norm = cm.astype("float") / cm.sum(axis=1, keepdims=True)
sns.heatmap(cm_norm, annot=True, fmt=".3f", cmap="Blues",
            xticklabels=["Real", "Fake"], yticklabels=["Real", "Fake"], ax=ax, cbar=False)
ax.set_title("Normalized Confusion Matrix", fontweight="bold")
ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
plt.tight_layout()
plt.savefig(os.path.join(CFG.PLOT_DIR, "05_confusion_matrix_normalized.png"),
            dpi=100, bbox_inches="tight")
plt.show()


# ============================================================================
# CELL 18 — Visualise correct & incorrect predictions on test set
# ============================================================================
correct_idx = np.where(y_true == y_pred)[0]
wrong_idx   = np.where(y_true != y_pred)[0]

def show_prediction_grid(indices, title, save_path):
    n = min(8, len(indices))
    if n == 0:
        return
    fig, axes = plt.subplots(2, 4, figsize=(14, 7))
    axes = axes.flatten()
    for i in range(n):
        row_idx = int(indices[i])
        img = Image.open(test_df.iloc[row_idx]["path"]).convert("RGB")
        true_lbl = "Real" if y_true[row_idx] == 0 else "Fake"
        pred_lbl = "Real" if y_pred[row_idx] == 0 else "Fake"
        conf = y_prob[row_idx] if y_pred[row_idx] == 1 else (1 - y_prob[row_idx])
        color = "green" if true_lbl == pred_lbl else "red"
        axes[i].imshow(img)
        axes[i].set_title(f"T:{true_lbl}  P:{pred_lbl}  ({conf:.2f})",
                          color=color, fontsize=10, fontweight="bold")
        axes[i].axis("off")
    for j in range(n, 8):
        axes[j].axis("off")
    plt.suptitle(title, fontsize=13, fontweight="bold")
    plt.tight_layout()
    plt.savefig(save_path, dpi=100, bbox_inches="tight")
    plt.show()

rng = np.random.default_rng(CFG.SEED)
if len(correct_idx) > 0:
    show_prediction_grid(
        rng.choice(correct_idx, size=min(8, len(correct_idx)), replace=False),
        "Correct Predictions (Test Set)",
        os.path.join(CFG.PLOT_DIR, "06_correct_predictions.png"),
    )
if len(wrong_idx) > 0:
    show_prediction_grid(
        rng.choice(wrong_idx, size=min(8, len(wrong_idx)), replace=False),
        "Incorrect Predictions (Test Set)",
        os.path.join(CFG.PLOT_DIR, "07_wrong_predictions.png"),
    )


# ============================================================================
# CELL 19 — Feature-extraction helper (for later cross-dataset use)
# ============================================================================
# The fine-tuned Swin-T can be used as a feature extractor.
# Swin-T's pooled feature dimension is 768.
#
# Standard timm idiom: build the same model with num_classes=0 (drops the head)
# and load the fine-tuned weights with strict=False (skips the head weights).

def build_feature_extractor(ckpt_path, device):
    """Return a Swin-T that outputs 768-dim feature vectors (no classifier)."""
    feat_model = timm.create_model(
        CFG.MODEL_NAME,
        pretrained=False,
        num_classes=0,   # head becomes Identity → model(x) returns pooled features
    )
    ckpt = torch.load(ckpt_path, map_location=device, weights_only=False)
    missing, unexpected = feat_model.load_state_dict(ckpt["model_state_dict"], strict=False)
    print(f"   Loaded backbone weights from {ckpt_path}")
    print(f"   Missing keys (expected: head.*): {len(missing)}")
    print(f"   Unexpected keys (expected: head.*): {len(unexpected)}")
    feat_model.eval().to(device)
    return feat_model


print("🧪  Demo: extracting features from a few test batches …")
feat_model = build_feature_extractor(best_ckpt_path, device)

demo_feats, demo_labels = [], []
with torch.no_grad():
    for i, (imgs, labels) in enumerate(test_loader):
        imgs = imgs.to(device, non_blocking=True)
        if device.type == "cuda":
            with autocast():
                feats = feat_model(imgs)
        else:
            feats = feat_model(imgs)
        demo_feats.append(feats.float().cpu().numpy())
        demo_labels.append(labels.numpy())
        if i >= 4:   # ~5 batches worth as a demo
            break

demo_feats  = np.concatenate(demo_feats, axis=0)
demo_labels = np.concatenate(demo_labels, axis=0)
print(f"   Feature matrix : {demo_feats.shape}   (n × feature_dim)")

np.save(os.path.join(CFG.OUTPUT_DIR, "demo_features.npy"), demo_feats)
np.save(os.path.join(CFG.OUTPUT_DIR, "demo_labels.npy"),   demo_labels)


# ============================================================================
# CELL 20 — Final summary
# ============================================================================
print("\n" + "=" * 70)
print("  📁  OUTPUT FILES  (in /kaggle/working/)")
print("=" * 70)
for root, _, files in os.walk(CFG.OUTPUT_DIR):
    for name in sorted(files):
        full = os.path.join(root, name)
        size_mb = os.path.getsize(full) / (1024 * 1024)
        rel = os.path.relpath(full, CFG.OUTPUT_DIR)
        print(f"   {rel:<45}  {size_mb:>8.2f} MB")

print("\n" + "=" * 70)
print("  🏆  FINAL RESULTS")
print("=" * 70)
print(f"  Best epoch      : {best_epoch}")
print(f"  Best val F1     : {best_val_f1:.4f}")
print(f"  Test accuracy   : {test_acc:.4f}")
print(f"  Test F1         : {test_f1:.4f}")
print(f"  Test ROC-AUC    : {test_auc:.4f}")
print("=" * 70)
print("✅  Done. Best model is at /kaggle/working/best_model.pth")
print("   Reuse it later as a feature extractor via build_feature_extractor().")


###############################################################################
#  HOW TO RELOAD LATER (in a new notebook / script)
#  ------------------------------------------------
#  import timm, torch
#  model = timm.create_model("swin_tiny_patch4_window7_224",
#                            pretrained=False, num_classes=2)
#  ckpt = torch.load("/path/to/best_model.pth",
#                    map_location="cpu", weights_only=False)
#  model.load_state_dict(ckpt["model_state_dict"])
#  model.eval()
#
#  # For feature extraction on a new dataset:
#  feat_model = timm.create_model("swin_tiny_patch4_window7_224",
#                                 pretrained=False, num_classes=0)
#  feat_model.load_state_dict(ckpt["model_state_dict"], strict=False)
#  feat_model.eval()      # → feat_model(x) returns 768-dim features
###############################################################################
```
